```python
nomes = ['Ana', 'Carlos', 'João', 'Maria' ]
print(nomes)

```

```python
#### Índice
nomes = ['Ana', 'Carlos', 'João', 'Maria' ]
print(nomes[0])
```

```python
##### Append
nomes = ['Ana', 'Carlos']
nomes.append('João')
print(nomes)
```

```python
#### Remove
nomes = ['Ana', 'Carlos', 'João', ]
nomes.remove('João')
nomes.remove('Carlos')
print(nomes)
```

```python
#### Pop
nomes = ['Ana', 'Carlos', 'João']
nomes.pop(2)
print(nomes)
```

```python
#### len
nomes = ['Ana', 'Carlos', 'João']
print(len(nomes))
```

```python
#### sort
nomes = ['João','Ana', 'Daniel', 'Anderson']
nomes.sort()
print(nomes)
```

```python
#### numeros
numeros = [3, 5, 6, 9, 5, 4, 2]
numeros.sort()
print(numeros)
```

```python
#### reverse
nomes = ['João','Ana', 'Daniel', 'Anderson']
nomes.reverse()
print(nomes)
```

```python
#### Alterando um valor
nomes = ['João','Ana', 'Daniel', 'Anderson']
nomes[1]='Maria'
print(nomes)
```

```python
####indentificar ou verificar
nomes = ['João','Ana', 'Daniel', 'Anderson']
print('Ana' in nomes )
```

```python
#### Atividade 01 — Gerenciamento de Projetos

### 01- ÍNDICE
projetos = ['ERP Corporativo', 'App Financeiro', 'Portal do Cliente', 'Sistema de Estoque', 'Dashboard BI']
print(projetos[0])

### 02 - APPEND
projetos = ['ERP Corporativo', 'App Financeiro', 'Portal do Cliente', 'Sistema de Estoque', 'Dashboard BI']
projetos.append('API de Integração')
print(projetos)

### 03 - REMOVE
projetos = ['ERP Corporativo', 'App Financeiro', 'Portal do Cliente', 'Sistema de Estoque', 'Dashboard BI', 'API de Integração']
projetos.remove('Sistema de Estoque')
print(projetos)

### 04 - POP
projetos = ['ERP Corporativo', 'App Financeiro', 'Portal do Cliente', 'Dashboard BI', 'API de Integração']
projetos.pop(1)
print(projetos)

### 05 - ÍNDICE ALTERADO
projetos = ['ERP Corporativo', 'Portal do Cliente', 'Dashboard BI', 'API de Integração']
projetos[2]= 'Dashboard Analytics'
print(projetos)

### 06 -
projetos = ['ERP Corporativo', 'Portal do Cliente', 'Dashboard Analytics', 'API de Integração']
print(len(projetos))
print(projetos)

### 07- SORT
projetos = ['ERP Corporativo', 'Portal do Cliente', 'Dashboard Analytics', 'API de Integração']
projetos.sort()
print(projetos)

#### 08 - REVERSE
projetos = ['API de Integração', 'Dashboard Analytics', 'ERP Corporativo', 'Portal do Cliente']
projetos.reverse()
print(projetos)

### 09 - IN
projetos = ['Portal do Cliente', 'ERP Corporativo', 'Dashboard Analytics', 'API de Integração']
print('ERP Corporativo' in projetos)
print(projetos)

### 10 - LISTE OS PROJETOS
projetos = ['Portal do Cliente', 'ERP Corporativo', 'Dashboard Analytics', 'API de Integração']
print(projetos)
```

```python
#### Atividade 02 — Gerenciamento de Chamados

#### 01 - ÍNDICE
chamados = ['Erro no Sistema', 'Acesso Bloqueado', 'Falha na Rede', 'Computador Lento', 'Problema no E-mail']
print(chamados[0])

#### 02 - APPEND
chamados = ['Erro no Sistema', 'Acesso Bloqueado', 'Falha na Rede', 'Computador Lento', 'Problema no E-mail']
chamados.append('Impressora Offline')
print(chamados)

#### 03 - REMOVE
chamados = ['Erro no Sistema', 'Acesso Bloqueado', 'Falha na Rede', 'Computador Lento', 'Problema no E-mail', 'Impressora Offline']
chamados.remove('Acesso Bloqueado')
print(chamados)

#### 04 - POP
chamados = ['Erro no Sistema', 'Falha na Rede', 'Computador Lento', 'Problema no E-mail', 'Impressora Offline']
print(chamados.pop(0))
print(chamados)

#### 05 - ÍNDICE ALTERADO
chamados = ['Falha na Rede', 'Computador Lento', 'Problema no E-mail', 'Impressora Offline']
chamados[3]= 'Computador sem Internet'
print(chamados)

#### 06 - LEN
chamados = ['Falha na Rede', 'Computador Lento', 'Problema no E-mail', 'Computador sem Internet']
print(len(chamados))
print(chamados)

#### 07 -
chamados = ['Falha na Rede', 'Computador Lento', 'Problema no E-mail', 'Computador sem Internet']
chamados.sort()
print(chamados)

#### 08 - REVERSE
chamados = ['Computador sem Internet', 'Computador Lento', 'Falha na Rede', 'Problema no E-mail']
chamados.reverse()
print(chamados)

#### 09 - IN
chamados = ['Problema no E-mail', 'Computador Lento', 'Falha na Rede', 'Computador sem Internet']
print('Erro no Sistema' in chamados)
print(chamados)

#### 10 - LISTE
chamados = ['Problema no E-mail', 'Computador Lento', 'Falha na Rede', 'Computador sem Internet']
print(chamados)
```
